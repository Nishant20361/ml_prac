Question: What is the difference between a feature and a target in machine learning?

Answer:
- Features are the information used to make a prediction.
- The target is what the model tries to predict.
Easy example: Imagine predicting a house’s price.
House size	Bedrooms	Location	Price
1,200 sq ft	2	Delhi	₹60 lakh
1,800 sq ft	3	Mumbai	₹1.2 crore


Here:
- Features: House size, Bedrooms, and Location
- Target: Price
