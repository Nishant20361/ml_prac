Question: What is the difference between supervised and unsupervised learning?
Answer:
- Supervised learning uses data containing correct answers, called labels.
- Unsupervised learning uses data without labels and tries to discover patterns.
Easy example:
- Predicting whether an email is spam using emails already labelled spam or not spam is supervised learning.
- Grouping customers by similar shopping habits without predefined groups is unsupervised learning.
Think of it like this:
- Supervised learning: learning with a teacher.
- Unsupervised learning: discovering patterns yourself.
