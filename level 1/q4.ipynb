Divide the Iris dataset into 80% training data and 20% testing data.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
```

```python
iris =load_iris()
```

```python
x= iris.data
y= iris.target
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,
    test_size=0.20,
    random_state=43)
```
